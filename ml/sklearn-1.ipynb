```python
from sklearn.linear_model import LinearRegression
```

```python
import numpy as np

X = np.array([[1],[2],[3],[4],[5]])   # features
y = np.array([2.1, 3.9, 6.2, 7.8, 10.1])  # targets
```

```python
print(X)
print(y)
```

```python
model = LinearRegression()
model.fit(X, y)           # learn
predictions = model.predict([[6], [7]])
print(predictions) 
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1],[2],[3],[4],[5]])
y = np.array([2.1, 3.9, 6.2, 7.8, 10.1])

model = LinearRegression()
model.fit(X, y)

X_pred = np.array([[6],[7]])
predictions = model.predict(X_pred)

X_line = np.linspace(0, 8, 100).reshape(-1, 1)
y_line = model.predict(X_line)

plt.figure(figsize=(7, 4))
plt.scatter(X, y, color='steelblue', zorder=5, label='Training data')
plt.scatter(X_pred, predictions, color='tomato', marker='*', s=150, zorder=5, label='Predictions (X=6,7)')
plt.plot(X_line, y_line, color='gray', linewidth=1.5, label=f'Fit: y = {model.coef_[0]:.2f}x + {model.intercept_:.2f}')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Regression')
plt.legend()
plt.tight_layout()
plt.show()

print(f"Slope: {model.coef_[0]:.4f}, Intercept: {model.intercept_:.4f}")
print(f"Predictions for X=6,7: {predictions}")
```
